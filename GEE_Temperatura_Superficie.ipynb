```python
import ee
import os
import geemap
import datetime
```

```python
ee.Authenticate()
ee.Initialize(project='zinc-fusion-290009')
```

```python
# Anos
startYear = 2001
endYear = 2024
```

```python
# =====================================================================================
# SEÇÃO 1: CONFIGURAÇÃO E FUNÇÕES BASE
# =====================================================================================
parelhas = ee.FeatureCollection('FAO/GAUL/2015/level2').filter(ee.Filter.equals('ADM0_NAME', 'Brazil')).filter(ee.Filter.equals('ADM1_NAME', 'Rio Grande Do Norte')).filter(ee.Filter.equals('ADM2_NAME', 'Parelhas'))
def maskL45789sr(image):
    cloudShadowBitMask = (1 << 3)
    cloudsBitMask = (1 << 5)
    qa = image.select('QA_PIXEL')
    mask = qa.bitwiseAnd(cloudShadowBitMask).eq(0).And(qa.bitwiseAnd(cloudsBitMask).eq(0))
    return image.updateMask(mask).select(['ST_B.*', 'QA_PIXEL']).copyProperties(image, ["system:time_start"])
def harmonizeAndToCelsius(image):
    tempK = image.select('ST_B.*').multiply(0.00341802).add(149.0)
    tempC = tempK.subtract(273.15)
    return tempC.rename('LST').copyProperties(image, ['system:time_start'])
l9 = ee.ImageCollection('LANDSAT/LC09/C02/T1_L2').select(['ST_B10', 'QA_PIXEL'])
l8 = ee.ImageCollection('LANDSAT/LC08/C02/T1_L2').select(['ST_B10', 'QA_PIXEL'])
l7 = ee.ImageCollection('LANDSAT/LE07/C02/T1_L2').select(['ST_B6', 'QA_PIXEL'])
l5 = ee.ImageCollection('LANDSAT/LT05/C02/T1_L2').select(['ST_B6', 'QA_PIXEL'])
landsatCollection = ee.ImageCollection(l5.merge(l7).merge(l8).merge(l9))
landsatLST = landsatCollection.filterDate(startDate, endDate).filterBounds(parelhas).map(maskL45789sr).map(harmonizeAndToCelsius)
```

```python
# =====================================================================================
#  PLOTAR A TEMPERATURA MÉDIA TOTAL NO MAPA
# =====================================================================================
print('Calculando e plotando a média total...')
Map = geemap.Map()
totalMeanLST = landsatLST.mean().clip(parelhas)
lstVisParams = {
  'min': 28,
  'max': 42,
  'palette': ['blue', 'cyan', 'green', 'yellow', 'red']
}
Map.centerObject(parelhas, 10)
Map.addLayer(totalMeanLST, lstVisParams, 'Temperatura Média Total (2001-2025)')
Map.addLayer(parelhas, {'color': 'black', 'fillColor': '00000000'}, 'Município de Parelhas')
display(Map)
```

```python
# --- 2.1: Exportar a Temperatura Média Total ---
print('Iniciando exportação da média TOTAL...')
totalMeanLST = landsatLST.mean().clip(parelhas)
task_total = ee.batch.Export.image.toDrive(
    image=totalMeanLST,
    description='LST_Media_Total_2001_2025',
    folder='GEE_Exports_Parelhas_Python',
    fileNamePrefix='LST_Media_Total_Parelhas_2001_2025',
    region=parelhas.geometry(),
    scale=30,
    fileFormat='GeoTIFF',
    maxPixels=1e10
)
task_total.start()
```

```python
# --- 2.2: Exportar as Temperaturas Médias Anuais
print(f'Iniciando exportações das médias ANUAIS de {startYear} a {endYear}...')
for y in range(startYear, endYear + 1):
    annualCollection = landsatLST.filter(ee.Filter.calendarRange(y, y, 'year'))
    annualMeanImage = annualCollection.mean().clip(parelhas)
    task_anual = ee.batch.Export.image.toDrive(image=annualMeanImage,description=f'Export_LST_Anual_{y}',folder='GEE_Exports_Parelhas_Python/Anual',fileNamePrefix=f'LST_Parelhas_Anual_{y}',region=parelhas.geometry(),scale=30,fileFormat='GeoTIFF',maxPixels=1e10)
    task_anual.start()
```

```python
# --- 2.3: Exportar as Temperaturas Médias Mensais
print(f'Iniciando exportações das médias MENSAIS de {startYear} a {endYear}...')
for y in range(startYear, endYear + 1):
    for m in range(1, 13):
        monthlyCollection = landsatLST.filter(ee.Filter.calendarRange(y, y, 'year')) \
                                        .filter(ee.Filter.calendarRange(m, m, 'month'))
        monthlyMeanImage = monthlyCollection.mean().clip(parelhas)
        task_mensal = ee.batch.Export.image.toDrive(
            image=monthlyMeanImage,
            description=f'Export_LST_Mensal_{y}_{m:02d}',
            folder='GEE_Exports_Parelhas_Python/Mensal',
            fileNamePrefix=f'LST_Parelhas_Mensal_{y}_{m:02d}',
            region=parelhas.geometry(),
            scale=30,
            fileFormat='GeoTIFF',
            maxPixels=1e10
        )
        task_mensal.start()
```
